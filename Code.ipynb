```python
import pandas as pd
import random
from pandasql import sqldf
```

Finance Database Simulator:

```python
day = []
genres = ["Food", "Transport", "Shopping", "Bills", "Entertainment"]
genres2 = {"Food":random.randint(3,5),
          "Transport":random.randint(2,3),
          "Shopping":random.randint(10,20),
          "Bills":random.randint(10,20),
          "Entertainment":random.randint(5,5)}
for month in (1,2,3):
    for days in range(1,30):
        r = random.randint(1,3)
        for d in range(r):
            choice = random.choice(genres)
            cash = genres2[choice]
            newrow = {"Day" : days, "Genre" : choice, "Cash" : cash, "Month" : month}
            day.append(newrow)
Finance = pd.DataFrame(day)
Finance["Cash"] = Finance["Cash"].astype(int)
print(Finance)
```

SQL:

```python
total = sqldf("""
SELECT SUM(Cash) AS TotalCash, ROUND(AVG(Cash),2) AS Mean
From Finance
       """)
month = sqldf("""
SELECT ROUND(SUM(Cash),2) AS Mean
FROM Finance
GROUP BY Month
""")
HighDays = sqldf("""
SELECT Day, Month, Cash
FROM Finance
GROUP BY Day, Month
ORDER BY Cash DESC
LIMIT 3
""")
highestgenre = sqldf("""
SELECT Genre, Sum(Cash) AS TotalCash
FROM Finance
GROUP BY Genre
ORDER BY TotalCash DESC
LIMIT 3;
""")
mostgenre = sqldf("""
SELECT Genre, COUNT(Cash) AS AmtofTransactions
FROM Finance
GROUP BY Genre
ORDER BY COUNT(Cash) DESC
LIMIT 3
""")
```

Result

```python
print(f"Total value for each month: \n${month}")
print(f"Mean amount of transactions per month: \n${month["Mean"].mean()}")
print(f"Total Spent = ${total["TotalCash"].sum()}, Average Spent Per Day = S{total["Mean"].sum()}")
print(f"Top Genre Spent = \n${highestgenre}")
print(f"Month and Day in which most month was spent: \n{HighDays}")
print(f"Genre with most amt of transactions: \n{mostgenre}")
```
